# Radar-DINO quickstart
Analyze one compatible radar NetCDF, inspect its feature and cluster, and save attention, UMAP, and t-SNE PNGs.

```python
%pip install "radar-dino[netcdf,hub,analysis,plot] @ git+https://github.com/DanielWefer/radar_dino.git@main"
```

```python
from pathlib import Path
from radar_dino import RadarDINO

NETCDF = Path("/path/to/KHTX_scan.nc")
OUTPUT = Path("radar_dino_output")
```

```python
dino = RadarDINO.from_pretrained(
    "dwefer/radar-dino-fieldtoken-v1",
    device="auto",
)
result = dino.analyze(NETCDF)
pngs = dino.save_plots(result, OUTPUT)
```

```python
print("feature shape:", result.feature.shape)
print("cluster:", result.cluster)
print("cluster probability:", result.cluster_probability)
print("UMAP:", result.umap)
print("t-SNE (display interpolation):", result.tsne)
print("neighbors:")
for neighbor in result.neighbors:
    print(neighbor["scan_id"], neighbor["similarity"])
```

```python
from IPython.display import display, Image

for name, path in pngs.items():
    print(name)
    display(Image(filename=str(path)))
```
